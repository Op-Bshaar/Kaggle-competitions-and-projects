```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, ParameterGrid
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.optimizers import Adam, SGD, RMSprop, Adagrad
```

```python
fashion_train = pd.read_csv('fashion-mnist_train.csv')
fashion_test = pd.read_csv('fashion-mnist_test.csv')
```

```python
plt.figure(figsize=(12,6))
for i in range(10):
  img = fashion_train.iloc[i, 1:].values.reshape(28,28)
  label_number = fashion_train.iloc[i]['label']
  plt.subplot(2,5,i+1)
  plt.imshow(img, cmap='gray')
  plt.title(label_number)
  plt.axis('off')
plt.show()
```

```python
X_train = fashion_train.iloc[:, 1:].values / 255.0
y_train = to_categorical(fashion_train['label'].values, 10)

X_test = fashion_test.iloc[:, 1:].values / 255.0
y_test = to_categorical(fashion_test['label'].values, 10)
```

```python
X_train_final, X_val, y_train_final, y_val = train_test_split(
    X_train, y_train, test_size=0.2, random_state=42
)
```

```python
def build_model(activation='relu', optimizer='adam', dropout_rate=0.2):
    model = Sequential([
        Dense(256, activation=activation, input_shape=(784,)),
        Dropout(dropout_rate),
        Dense(128, activation=activation),
        Dropout(dropout_rate),
        Dense(10, activation='softmax')
    ])
    model.compile(optimizer=optimizer,
                  loss='categorical_crossentropy',
                  metrics=['accuracy'])
    return model
```

```python
activations = ['relu', 'tanh', 'sigmoid', 'swish']
optimizers = ['adam', 'sgd', 'rmsprop', 'adagrad']

param_grid = {'activation': activations, 'optimizer': optimizers}
grid = list(ParameterGrid(param_grid))
```

```python
results = []

for params in grid:
    print(f"Training model with {params}")
    model = build_model(params['activation'], params['optimizer'])
    history = model.fit(X_train_final, y_train_final,
                        epochs=5, batch_size=32,
                        validation_data=(X_val, y_val),
                        verbose=3)
    val_acc = history.history['val_accuracy'][-1]
    results.append({**params, 'val_accuracy': val_acc})

df_results = pd.DataFrame(results)
print(df_results)
```

```python
pivot_df = df_results.pivot(index='activation', columns='optimizer', values='val_accuracy')

plt.figure(figsize=(10,7))
sns.heatmap(pivot_df, annot=True, fmt=".4f", cmap='coolwarm', linewidths=.5)
plt.title("Validation Accuracy: Activation vs Optimizer")
plt.show()
```

```python
best_params = df_results.loc[df_results['val_accuracy'].idxmax()]
print("Best params:", best_params)

best_model = build_model(best_params['activation'], best_params['optimizer'])
best_model.fit(X_train_final, y_train_final, epochs=5, batch_size=32,
               validation_data=(X_val, y_val), verbose=0)

test_loss, test_acc = best_model.evaluate(X_test, y_test, verbose=0)
print("Final Test Accuracy:", test_acc)
```
